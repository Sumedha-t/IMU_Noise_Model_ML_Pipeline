# 01 — Data Inspection

## Objective

Inspect the IPIN 2024 Track 4 IMU dataset before performing
noise characterization or machine-learning modelling.

## Questions

1. What variables are available?
2. What are their units?
3. What is the sampling frequency?
4. Are there missing or invalid values?
5. What does the raw sensor signal look like?

```python
import pandas as pd
accel = pd.read_csv(r'D:\PROJECTS\YAANENDRIYA\data\raw\acceleration_Allan.csv', header = None, names = ['time', 'x', 'y', 'z'])
```

```python
print(accel.head())
print(accel.columns)
print(accel.shape)
```

```python
dt = accel['time'].diff()
average_dt = dt.mean()
sampling_frequency = 1 / average_dt
print(dt.head())
print(average_dt)
print(sampling_frequency)
info_dt = dt.describe()
print(info_dt)
gap_dt = dt[dt > 0.005]
print(gap_dt)
```
